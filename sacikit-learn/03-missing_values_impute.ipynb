```python
import pandas as pd 
import numpy as np 
import seaborn as sns 
%matplotlib inline
from sklearn.impute import KNNImputer
from sklearn.experimental import enable_iterative_imputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from sklearn.model_selection import train_test_split
```

```python
data = sns.load_dataset('titanic')
data.head()
```

```python
# check the number of missing values
data.isnull().sum().sort_values(ascending=False)
```

```python
# how to drop the column
data.drop('deck', axis=1, inplace=True)
data.head()
```

```python
# impute missing values with mean 
data['age'] = data['age'].fillna(data['age'].mean())
data.isnull().sum().sort_values(ascending=False)
```

```python
# impute missing values with median 
data['age'] = data['age'].fillna(data['age'].median())
data.isnull().sum().sort_values(ascending=False)
```

```python
data['embarked'] = data['embarked'].fillna(data['embarked'].mode()[0])
# embark_town try it with this words
data.isnull().sum().sort_values(ascending=False)
```

```python
from sklearn.impute import KNNImputer

Imputer = KNNImputer(n_neighbors=4)

data['age'] = Imputer.fit_transform(data[['age']])
data.isnull().sum().sort_values(ascending=False)
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

df = sns.load_dataset('titanic')
df.head()
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

columns_to_encode = ['sex','embarked','who','class','embark_town','alive']

labels_encoder = {}

for col in columns_to_encode:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    labels_encoder[col] = le
df.head()
```

```python
for col in columns_to_encode:
    le = labels_encoder[col]
    df[col] = le.inverse_transform(df[col].astype(int))
df.head()
    
```

```python
imputer = IterativeImputer(max_iter=10)

columns_to_impute = ['age','embarked','embark_town','deck']

for col in columns_to_impute:
    df[col] = imputer.fit_transform(df[[col]])
df.isnull().sum().sort_values(ascending=False)
    
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from sklearn.impute import SimpleImputer

df = sns.load_dataset('titanic')
df.head()

df.drop('deck', axis=1, inplace=True)
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

labels_encoder = {}

for col in columns_to_encode:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    labels_encoder[col] = le
df.head()
```

```python
# splie the data into 2 parts, onw for missing and one witout missing

df_with_missing = df[df['age'].isna()]
df_without_missing = df.dropna()
```

```python
# print(df.shape)
# print(df_with_missing.shape)
# print(df_without_missing.shape)
df_with_missing.head()
```

```python
X = df_without_missing.drop(['age'], axis=1)
Y = df_without_missing['age']

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.20, random_state=42)

r_f_model = RandomForestRegressor(n_estimators=100, random_state=42)
r_f_model.fit(X,Y)

Y_pred = r_f_model.predict(X_test)

print(f"Mean squared error: {mean_squared_error(Y_test, Y_pred)}")
print(f"r2_score: {r2_score(Y_test, Y_pred)}")
print(f"mean_absolute_error: {mean_absolute_error(Y_test, Y_pred)}")

# with numpy
print(f"mean_absolute_error: {np.sqrt(mean_absolute_error(Y_test, Y_pred))}")
```
