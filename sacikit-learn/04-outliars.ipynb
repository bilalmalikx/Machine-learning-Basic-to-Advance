```python
import pandas as pd
import numpy as np
import seaborn as sns

data = pd.DataFrame({'Age': [11, 12, 13, 14, 15, 16, 17, 18, 10, 20, 90],})
```

```python
data.head()
```

```python
sns.histplot(data['Age'], bins=5, kde=True)
```

```python
data
```

```python
data['mean'] = np.mean(data['Age'])
data['std'] = np.std(data['Age'])

data['Z-Score'] = (data['Age'] - data['mean']) / data['std']
data
```

```python
data = data[data['Z-Score'] <= 3]
print(f"here is the filtered data:\n{data}")
```

```python
from scipy import stats

data = [2.5,2.7,2.8,3.0,3.1,3.2,3.3,3.4,3.5,110.0]

z_scores = np.abs(stats.zscore(data))

# set the theardshort

theardshort = 2.5
outliars = np.where(z_scores > theardshort)[0]
print(f"data are at index: {data}")
print(f"outliars are at index: {outliars}")

data = [data[i] for i in range(len(data)) if i not in outliars]
print(f"filtered data: {data}")
```
